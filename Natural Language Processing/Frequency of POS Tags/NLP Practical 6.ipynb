```python
import nltk
from collections import Counter
nltk.download('averaged_perceptron_tagger_eng')

text = """
Mark Grayson was just an ordinary high school student living a seemingly normal life, 
until everything changed. As the son of Nolan Grayson, secretly known as the world's 
most powerful superhero, Omni-Man, Mark always knew he would eventually inherit 
extraordinary abilities. Shortly after his seventeenth birthday, his powers finally 
manifested, granting him super strength, incredible speed, enhanced durability, and 
the exhilarating ability to fly. Eager to follow in his father's legendary footsteps, 
Mark adopted the superhero moniker Invincible and began his grueling training. 
However, balancing his demanding new responsibilities with the everyday struggles of 
teenage life, such as homework, awkward romances, and part-time jobs, proved to be 
overwhelmingly difficult. Furthermore, as Mark delved deeper into the complex and 
often brutal reality of being a hero, he quickly discovered that his father's legacy, 
and the universe itself, held dark secrets he was entirely unprepared to face.
"""

tokens = nltk.word_tokenize(text)

pos_tags = nltk.pos_tag(tokens)

nouns = [word for word, tag in pos_tags if tag.startswith('NN')]

tags_only = [tag for word, tag in pos_tags]
pos_frequency = dict(Counter(tags_only))


print("--- Extracted Nouns ---")
unique_nouns = sorted(list(set(nouns)))
print(unique_nouns) 
print(f"\nTotal Nouns found (including duplicates): {len(nouns)}")

print("\n--- POS Tag Frequency Dictionary ---")
for tag, freq in sorted(pos_frequency.items(), key=lambda x: x[1], reverse=True):
    print(f"'{tag}': {freq}")
```
