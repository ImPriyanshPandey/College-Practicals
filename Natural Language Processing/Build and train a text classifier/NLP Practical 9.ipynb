```python
import tensorflow as tf
import numpy as np

# 1. Prepare the Data
training_sentences = [
    "I absolutely love this product",
    "This is the worst thing I have ever bought",
    "Amazing quality and fast shipping",
    "Terrible customer service and broken item",
    "Highly recommend this to everyone",
    "Do not waste your money on this"
]
training_labels = np.array([1, 0, 1, 0, 1, 0])

testing_sentences = [
    "I really like this",
    "This is awful and bad"
]

# FIX: Convert the text lists to TensorFlow string tensors instead of NumPy arrays 
# to avoid the 'Invalid dtype' error in Keras 3.
X_train = tf.constant(training_sentences, dtype=tf.string)
X_test = tf.constant(testing_sentences, dtype=tf.string)

# 2. Text Vectorization
vocab_size = 1000
max_length = 10

vectorize_layer = tf.keras.layers.TextVectorization(
    max_tokens=vocab_size,
    output_mode='int',
    output_sequence_length=max_length
)

# Adapt the layer to the vocabulary
vectorize_layer.adapt(training_sentences)

# 3. Build the Keras Model
model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,), dtype=tf.string),
    vectorize_layer,
    tf.keras.layers.Embedding(input_dim=vocab_size, output_dim=16),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# 4. Compile the Model
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()

# 5. Train the Model
print("\n--- Starting Training ---")
model.fit(
    X_train, # Using the fixed tensor here!
    training_labels, 
    epochs=20, 
    verbose=1
)

# 6. Make Predictions
print("\n--- Making Predictions ---")
predictions = model.predict(X_test)

for i, text in enumerate(testing_sentences):
    sentiment = "Positive" if predictions[i][0] > 0.5 else "Negative"
    confidence = predictions[i][0]
    print(f"Text: '{text}' | Predicted: {sentiment} (Score: {confidence:.4f})")
```
