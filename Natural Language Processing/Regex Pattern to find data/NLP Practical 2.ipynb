```python
import re

text = """
Contact us at vk2441013@gmail.com .
Call +91-9315195409 or 8851721808.
Event date: 13/08/2026 .
Follow us on #Yeagerists #Rumbling #AttackonTitan
"""

email_usernames = re.findall(r'([a-zA-Z0-9._%+-]+)@', text)

hashtags = re.findall(r'#\w+', text)

dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b|\b\d{4}-\d{2}-\d{2}\b', text)

phones = re.findall(r'\+?\d{2}[- ]?\d{10}|\b\d{10}\b', text)

print("Email Usernames:", email_usernames)
print("Hashtags:", hashtags)
print("Dates:", dates)
print("Phone Numbers:", phones)
```
