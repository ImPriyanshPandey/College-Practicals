```python
import nltk
from collections import Counter

corpus = [ """
Erwin Smith serves as the stoic and brilliantly tactical thirteenth Commander of the 
Survey Corps in the grim universe of Attack on Titan. Renowned for his unparalleled 
strategic mind and commanding presence, he leads humanity's desperate fight against 
the terrifying, man-eating Titans. Erwin is not merely a soldier; he is a visionary 
who carries the crushing weight of countless casualties on his broad shoulders. Driven 
by an insatiable, deep-seated childhood curiosity to uncover the hidden truth about 
the walled world, he is willing to sacrifice anything, including his own humanity and 
the lives of his devoted comrades, to achieve his ultimate goal. His legendary, 
impassioned speeches possess the extraordinary power to inspire terrified soldiers to 
ride willingly toward their certain death. Despite the controversies surrounding his 
ruthless gambling on human lives, Erwin remains an indispensable symbol of hope and 
unwavering resolve in a remarkably cruel, unforgiving world, cementing his legacy.
"""
]

tokens = []
for sentence in corpus:
    tokens.extend(sentence.lower().split())

vocab = set(tokens)
V = len(vocab)

unigram_counts = Counter(tokens)

bigrams = list(nltk.bigrams(tokens))
bigram_counts = Counter(bigrams)

N = len(tokens)

# Unigram probability with add-one smoothing
def unigram_prob(word):
    return (unigram_counts[word] + 1) / (N + V)

# Bigram probability with add-one smoothing
def bigram_prob(w1, w2):
    return (bigram_counts[(w1, w2)] + 1) / (unigram_counts[w1] + V)

# Compute sentence probability using bigram model
def sentence_probability(sentence):
    words = sentence.lower().split()
    prob = 1
    
    for i in range(len(words)-1):
        prob *= bigram_prob(words[i], words[i+1])
    
    return prob

test_sentence = "Commander Erwin is the greatest commander of human history"
print("Sentence:", test_sentence)
print("Sentence Probability:", sentence_probability(test_sentence))
```
