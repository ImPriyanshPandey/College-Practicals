```python
import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('maxent_ne_chunker_tab')

headlines = [
    "Attack on Titan Finale Breaks Streaming Records Worldwide"
"Jujutsu Kaisen Season 3 Officially Announced by MAPPA"
"Demon Slayer Movie Premieres in Tokyo With Huge Fan Turnout"
"One Piece Celebrates 25 Years With Special Event in Osaka"
"Netflix Announces New Live-Action Adaptation of Naruto"
"Crunchyroll Expands Anime Library for Fans in India"
"My Hero Academia Final Season Trailer Released by Studio Bones"
"Chainsaw Man Wins Anime of the Year at Crunchyroll Anime Awards"
"Spy x Family Movie Announced After Global Success"
"Dragon Ball Exhibition Opens at Museum in Kyoto"
]

for headline in headlines:
    print("\nHeadline:", headline)

    words = nltk.word_tokenize(headline)
    
    pos_tags = nltk.pos_tag(words)

    named_entities = nltk.ne_chunk(pos_tags)

    print("Named Entities:")
    for chunk in named_entities:
        if hasattr(chunk, 'label'):  
            entity = " ".join(c[0] for c in chunk)
            print(f"{entity} -> {chunk.label()}")
```
